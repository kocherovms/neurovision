# Annealers

## const

```python
steps_count = 1000
annel = get_anneal('const', 0.00025)
lrs = []

for s in range(steps_count):
    lr = annel(s / steps_count)
    lrs.append(lr)

plt.grid()
plt.plot(np.arange(steps_count), lrs);
```
