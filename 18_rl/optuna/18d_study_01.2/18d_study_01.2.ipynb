# set_hyperparameters

```python
# @launchit.collect
def set_hyperparameters(HP, optuna_study, optuna_trial):
    import random
    HP.system.comment = None
    HP.system.random_seed = random.randint(0, 100)
    HP.system.is_torch_deterministic = True
    HP.system.is_torch_compile = True
    HP.system.use_amp = True
    
    HP.dataset.train = [
        'train_dataset:100',
        'train_dataset:101',
        'train_dataset:102',
        'train_dataset:103',
        'train_dataset:104',
        'train_dataset:105',
        'train_dataset:106',
        'train_dataset:107',
        'train_dataset:108',
        'train_dataset:109',
    ]
    HP.dataset.test = 'test_dataset:2'
    
    parents = [
        '18d_world_model_02:8',
        '18d_world_model_02:17',
    ]
    parent_ind = random.randint(0, len(parents)-1)
    HP.model.parent = parents[parent_ind]
    HP.model.is_reconstruction = 'from parent'
    HP.model.is_prediction = 'from parent'
    HP.model.sequence_length = 'from parent'
    HP.model.actions_count = 'from parent'
    HP.model.ob_shape = 'from parent'
    HP.model.vision_head = 'from parent'
    HP.model.g_encoding = 'from parent'
    HP.model.d_model = 'from parent'
    HP.model.layers_count = 'from parent'
    HP.model.heads_count = 'from parent'
    HP.model.attention_backend = 'from parent'
    HP.model.render_heads_count = 'from parent'
    
    HP.train.epochs_count = 100
    HP.train.batch_size = 128
    HP.train.optimizer = 'AdamW'
    HP.train.max_grad_norm = 1.0
    HP.train.learn_rate = 'const(0.00025)'
    HP.train.bce_loss_coef = 'const(1.0)'
    HP.train.edge_loss_coef = 'const(1.0)'
    
    return HP
# @launchit.stop
```
