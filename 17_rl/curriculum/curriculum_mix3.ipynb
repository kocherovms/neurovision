```python
# @launchit.collect
from collections import namedtuple
import itertools
import torch

class Curriculum:
    PlanItem = namedtuple('PlanItem', 'game_modifs game_modifs_probs env_game_modifs')
    
    def get_plan_item(self, t, envs_count):
        env_game_modifs = dict(map(lambda i: (i, ['reset_last_life']), range(envs_count)))

        return Curriculum.PlanItem(
            game_modifs=None,
            game_modifs_probs=torch.tensor([1]),
            env_game_modifs=env_game_modifs,
        )

    def get_game_modifs_set(self):
        return set(['reset_last_life'])
```

```python
curriculum = Curriculum()
curriculum.get_plan_item(0, 5)
curriculum.get_game_modifs_set()
```
